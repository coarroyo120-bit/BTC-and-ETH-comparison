## Analysis

In this section, we compare Bitcoin (BTC) and Ethereum (ETH) across price trends, daily returns, volatility, and average monthly performance.

## Price Trend Comparison
```python
plt.figure(figsize=(12,6))
plt.plot(btc.index, btc["Close"], label="BTC", color="orange")
plt.plot(eth.index, eth["Close"], label="ETH", color="purple")
plt.title("BTC vs ETH Closing Price - Last 12 Months")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()
plt.show()
```
📊 **Insight:**  
Bitcoin maintains a higher absolute price compared to Ethereum, but both follow similar market cycles with strong correlation in upward and downward moves.

## Daily Returns
```python
plt.figure(figsize=(12,6))
plt.hist(btc["Daily Return"].dropna(), bins=50, alpha=0.5, label="BTC")
plt.hist(eth["Daily Return"].dropna(), bins=50, alpha=0.5, label="ETH")
plt.title("BTC vs ETH Daily Returns Distribution")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
📊 **Insight:**  
Ethereum’s distribution is wider, showing **higher volatility** compared to Bitcoin.  
This suggests ETH experiences more frequent large swings (both gains and losses).

## Most Volatile Days
```python
btc_vol_day = btc["Daily Return"].abs().idxmax()
btc_vol_value = float(btc.loc[btc_vol_day, "Daily Return"])

eth_vol_day = eth["Daily Return"].abs().idxmax()
eth_vol_value = float(eth.loc[eth_vol_day, "Daily Return"])

print(f"Most volatile BTC day: {btc_vol_day.date()} with return {btc_vol_value:.2%}")
print(f"Most volatile ETH day: {eth_vol_day.date()} with return {eth_vol_value:.2%}")
```
📊 **Insight:**  
- The most volatile BTC day occurred on **2024-11-11** with a return of **10.22%**.  
- The most volatile ETH day occurred on **2025-05-08** with a return of **21.80%**.  
Ethereum’s swings were more extreme than Bitcoin’s.

## Monthly Average Returns
```python
btc_monthly = btc["Daily Return"].resample("M").mean()
eth_monthly = eth["Daily Return"].resample("M").mean()

plt.figure(figsize=(12,6))
plt.plot(btc_monthly.index, btc_monthly, label="BTC Monthly Avg Return", marker="o")
plt.plot(eth_monthly.index, eth_monthly, label="ETH Monthly Avg Return", marker="o")
plt.axhline(0, color="black", linestyle="--")
plt.title("BTC vs ETH Average Monthly Returns")
plt.xlabel("Month")
plt.ylabel("Average Return")
plt.legend()
plt.show()
```
📊 **Insight:**  
Ethereum outperformed Bitcoin in several months but also experienced more negative months.  
This reinforces ETH’s higher **risk/reward profile**.
